```python
a = np.array([10, 20, 30])
```

```python
print("1", a.shape)
print("2", a[:, None])
print("3", a[:, None].shape)
print("4", a[None, :])
print("5", a[None, :].shape)
```

```python
col = np.array([1, 2, 3])[:, None]     # shape (3, 1)
row = np.array([10, 20, 30, 40])[None, :]  # shape (1, 4)
```
